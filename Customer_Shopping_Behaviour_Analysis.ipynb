```python
import os
import tkinter as tk
from tkinter import filedialog
import pandas as pd
from sqlalchemy import create_engine

# ==========================================
# STEP 1: CHOOSE AND LOAD FILE
# ==========================================
root = tk.Tk()
root.withdraw()
root.attributes('-topmost', True)

print("Opening file browser... Please select your dataset file.")
file_path = filedialog.askopenfilename(
    title="Select your dataset file",
    filetypes=[("CSV and Excel files", "*.csv *.xlsx *.xls")]
)

if not file_path:
    print("No file was selected!")
else:
    print(f"\nFile selected: {file_path}")
    
    # Load dataset
    if file_path.endswith('.csv'):
        df = pd.read_csv(file_path)
    else:
        df = pd.read_excel(file_path)
        
    print("\nInitial dataset loaded successfully.")

    # ==========================================
    # STEP 2: YOUR EXACT DATA CLEANING STEPS
    # ==========================================
    print("\n--- Cleaning and Transforming Data ---")

    # 1. Fill missing Review Rating with category median
    if 'Review Rating' in df.columns and 'Category' in df.columns:
        df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))

    # 2. Clean column names (lowercase & replace spaces with underscores)
    df.columns = df.columns.str.lower().str.replace(' ', '_')

    # 3. Rename purchase amount column
    if 'purchase_amount_(usd)' in df.columns:
        df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})

    # 4. Create 'aged_group' column
    if 'age' in df.columns:
        labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
        df['aged_group'] = pd.qcut(df['age'], q=4, labels=labels)

    # 5. Create 'purchase_frequency_days' mapping column
    if 'frequency_of_purchases' in df.columns:
        frequency_mapping = {
            'Fortnightly': 14,
            'Weekly': 7,
            'Monthly': 30,
            'Quarterly': 90,
            'Bi-Weekly': 14,
            'Annually': 365,
            'Every 3 Months': 90
        }
        df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)

    # 6. Drop redundant 'promo_code_used' column if present
    if 'promo_code_used' in df.columns:
        df = df.drop('promo_code_used', axis=1)

    print("Cleaned Columns:", list(df.columns))
    print(f"Cleaned Shape: {df.shape}")

    # ==========================================
    # STEP 3: EXPORT CLEANED DATA TO POSTGRESQL
    # ==========================================
    print("\n--- Exporting Cleaned Data to PostgreSQL ---")
    
    # Update your database credentials
    username = "postgres"        
    password = "Deekshi2005"   # <--- Enter your actual PostgreSQL password here
    host = "localhost"
    port = "5432"
    database = "customer_behaviour"

    # Connect & Export
    engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")
    
    # Save clean df to table 'customer'
    df.to_sql("customer", engine, if_exists="replace", index=False)

    print("\nSUCCESS! Cleaned dataset exported to PostgreSQL table 'customer'.")
```
